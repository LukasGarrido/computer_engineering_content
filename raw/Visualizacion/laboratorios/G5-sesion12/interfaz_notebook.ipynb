```python
# --- Función de Ayuda: NumPy (OpenCV) a Tkinter Image ---
def np_to_tkimage(np_image):
    """
    Convierte un array de NumPy (OpenCV) a una imagen compatible con Tkinter.
    """
    if len(np_image.shape) == 2: # Escala de grises
        img_pil = Image.fromarray(np_image)
    else: # Color
        img_pil = Image.fromarray(cv2.cvtColor(np_image, cv2.COLOR_BGR2RGB))
    
    return ImageTk.PhotoImage(image=img_pil)


# --- Función de Ayuda: Lógica de PCA ---
def aplicar_pca(imagen_np, k):
    """
    Aplica PCA y reconstruye una imagen en escala de grises.
    """
    # 1. Normalizar (Centrar la media) - Convertir a float para precisión
    img_norm = imagen_np.astype(float) - np.mean(imagen_np, axis=0)
    
    # 2. Aplicar PCA
    pca = PCA(n_components=k)
    transformed = pca.fit_transform(img_norm)
    
    # 3. Reconstruir
    reconstructed = pca.inverse_transform(transformed)
    
    # 4. Devolver a la escala original (sumar la media)
    reconstructed_img = reconstructed + np.mean(imagen_np, axis=0)
    
    # 5. Asegurar rango de píxeles [0, 255] y tipo uint8
    reconstructed_img = np.clip(reconstructed_img, 0, 255).astype(np.uint8)
    
    return reconstructed_img


# --- Clase Principal de la Aplicación ---
class ImageProcessorApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Procesador de Imágenes (EIN095B)")
        
        # --- Atributos de Estado ---
        self.original_image_np = None  # Imagen en formato NumPy (para procesamiento)
        self.original_image_tk = None  # Imagen en formato Tkinter (para display)
        self.processed_image_tk = None # Imagen procesada en formato Tkinter
        
        # --- Configurar la GUI ---
        
        # 1. Frame de controles (superior)
        control_frame = tk.Frame(self.root, pady=10)
        control_frame.pack(side=tk.TOP, fill=tk.X, padx=10, pady=5)

        # Botón de Carga (Actividad 1)
        self.load_button = tk.Button(control_frame, text="Cargar Imagen", command=self.cargar_imagen)
        self.load_button.pack(side=tk.LEFT)

        # --- ### TAREA: ACTIVIDAD 2 (WIDGETS) ### ---
        # 1. Crear self.filter_var (un tk.StringVar)
        # 2. Crear un tk.OptionMenu para los filtros (Usar self.filter_var)
        # 3. Crear un tk.Button "Aplicar Filtro" (llamar a self.aplicar_filtro)
        # ... (Añadir widgets al control_frame usando .pack())
        
        # --- ### TAREA: ACTIVIDAD 3 (WIDGETS) ### ---
        # 1. Crear self.pca_slider (un tk.Scale horizontal)
        # 2. Crear un tk.Button "Reconstruir PCA" (llamar a self.reconstruir_pca)
        # ... (Añadir widgets al control_frame usando .pack())
        
        # 2. Frame de visualización (principal)
        display_frame = tk.Frame(self.root)
        display_frame.pack(fill=tk.BOTH, expand=True)
        
        # Panel Izquierdo (Original)
        self.original_label = tk.Label(display_frame, text="Imagen Original\n(Cargue una imagen para comenzar)", relief="sunken", padx=10, pady=10)
        self.original_label.pack(side=tk.LEFT, fill=tk.BOTH, expand=True, padx=10, pady=10)
        
        # Panel Derecho (Procesada)
        self.processed_label = tk.Label(display_frame, text="Imagen Procesada", relief="sunken", padx=10, pady=10)
        self.processed_label.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True, padx=10, pady=10)

    # --- Métodos de la GUI (Callbacks) ---

    def cargar_imagen(self):
        # --- ### TAREA: ACTIVIDAD 1 (LÓGICA) ### ---
        # 1. Pedir al usuario el path del archivo (usar filedialog.askopenfilename)
        # 2. Leer la imagen con cv2.imread(..., cv2.IMREAD_GRAYSCALE)
        # 3. Almacenar en self.original_image_np
        # 4. Convertir a formato Tkinter (usar np_to_tkimage)
        # 5. Configurar self.original_label (usar .config(image=...))
        # 6. Guardar la referencia en self.original_image_tk
        # 7. Repetir 5 y 6 para self.processed_label y self.processed_image_tk
        # 8. (Opcional) Actualizar el rango del slider de PCA
        
        print("Función 'cargar_imagen' llamada. ¡Completa esta parte!")
        pass


    def aplicar_filtro(self):
        # --- ### TAREA: ACTIVIDAD 2 (LÓGICA) ### ---
        # 1. Revisar si self.original_image_np es None. Si es así, 'return'.
        # 2. Obtener el filtro seleccionado de self.filter_var.get()
        # 3. Definir los kernels (Blur, Laplace)
        # 4. Aplicar cv2.filter2D según el filtro seleccionado
        # 5. Convertir la imagen resultante a Tkinter (np_to_tkimage)
        # 6. Actualizar self.processed_label y self.processed_image_tk
        
        print("Función 'aplicar_filtro' llamada. ¡Completa esta parte!")
        pass


    def reconstruir_pca(self):
        # --- ### TAREA: ACTIVIDAD 3 (LÓGICA) ### ---
        # 1. Revisar si self.original_image_np es None. Si es así, 'return'.
        # 2. Obtener el valor 'k' de self.pca_slider.get()
        # 3. Llamar a la función aplicar_pca(self.original_image_np, k)
        # 4. Convertir la imagen resultante a Tkinter (np_to_tkimage)
        # 5. Actualizar self.processed_label y self.processed_image_tk

        print("Función 'reconstruir_pca' llamada. ¡Completa esta parte!")
        pass


# --- Punto de Entrada Principal ---
# No usamos 'if __name__ == "__main__"' en notebooks, 
# ya que queremos que se ejecute directamente.

print("Iniciando aplicación Tkinter...")
print("¡Cierra la ventana de la aplicación para que la celda termine de ejecutarse!")

root = tk.Tk()
root.geometry("1000x700") # Tamaño inicial
app = ImageProcessorApp(root)

# Esta línea bloqueará la ejecución de la celda hasta que la ventana se cierre.
root.mainloop()

print("Aplicación cerrada. Kernel liberado.")
```
